```python
import pandas as pd

#Read the csv file and store in dataframe
df=pd.read_csv("/content/city_weather_data.csv")
print(df)

#Display only first 5 rows
print(df.head())

#Display no. of rows,columns,datatypes etc.
print(df.info())
```

```python
#Display the rows and colummns.
print("shape(rows,cols):",df.shape)

print("\nColumns names:",list(df.columns))

#Display the datatype
print("\ndata types per column:")
print(df.dtypes)

#Display count,mean,median,std etc.
print("\nQuick summary stats (numeric columns only):")
df.describe()
```

```python
print("Row with label 2(via.loc):")
print(df.loc[3])

print("\nFirst 2 rows,first 2 columns(via.iloc):")
print(df.iloc[0:3, 0:3])

print("\nJust the 'city' column(return the series):")
print(df["city"])

print("\nSpecific cell:city of row 3(via.loc):")
print(df.loc[5,"city"])
```

```python
#Filtering rows with boolean conditions
#A boolean condition on a column produces a series of True/False
#which we then use to mask

hot_cities=df[df["temprature_c"]>38]  #single condition
print("Cities hotter than 38 C:")
print(hot_cities)

#combine multiple conditions with & (and) /| (or)
costal_and_humid=df[(df["is_coastal"]==True) & (df["humidity_pct"]>75)]
print("\nCoastal cities with humidity >75%:")
print(costal_and_humid)
```

```python

df=pd.read_csv("/content/employee_messy_data.csv")
print(df)
```

```python
#handle missing values
clean=df.copy()

#fill missing 'age' with the column's median
clean["age"]=clean["age"].fillna(clean["age"].median())

#fill missing 'name' with a placeholder
clean["name"]=clean["name"].fillna("Unknown")

print("After filling missing values:")
clean
```

```python
#detect and remove duplicate rows
print("Duplicate rows(based on employee_id):")
print(clean[clean.duplicated(subset="employee_id",keep=False)])

clean=clean.drop_duplicates(subset="employee_id",keep="first").reset_index(drop=True)
print("\nAfter removing duplicates:")
clean
```

```python
print("Dtypes before cleaning:")
print(clean.dtypes)

# Fill NaN values in 'Salary' with the median before converting to int
clean["Salary"] = clean["Salary"].fillna(clean["Salary"].median()).astype(int)
clean["join_date"]=pd.to_datetime(clean["join_date"])

clean["age"]=clean["age"].astype(int)
print("\nDtypes after cleaning:")
print(clean.dtypes)
clean
```

```python

df=pd.read_csv("/content/sales_data.csv")
print(df)
```

```python
#groupby('region')
Category_summary=df.groupby("Category").agg(
    total_revenue=("revenue","sum"),
    avg_units_sold=("units_sold","mean"),
    num_transactions=("revenue","count"),

).reset_index()

print("Revenue summary by Category:")
Category_summary
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

sns.set_theme(style="whitegrid")

df = pd.read_csv("ecommerce_orders_dataset - ecommerce_orders_dataset.csv")

# HISTOGRAM: Distribution of Order Amount
plt.figure(figsize=(8,5))

sns.histplot(df["Tax_Amount"], bins=30, kde=True)

plt.title("Distribution of Order Amount", fontsize=14, fontweight="bold")
plt.xlabel("Tax Amount ($)")
plt.ylabel("Frequency")
```

```python
#BAR PLOT: Average Order Amount by Warehouse Region
plt.figure(figsize=(10, 6))
sns.barplot(x="Warehouse_Region", y="Order_Amount", data=df)

plt.xticks(rotation=0) # rotation=90 if region names overlap, but 0 is clean here
plt.title("Average Order Amount by Warehouse Region", fontsize=14, fontweight="bold")
plt.xlabel("Warehouse Region", fontsize=12)
plt.ylabel("Average Order Amount ($)", fontsize=12)
plt.show()
```

```python
# SCATTER PLOT: Unit Price vs. Order Amount
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df, x="Unit_Price", y="Order_Amount", alpha=0.6, color="blue"
)

plt.title("Unit Price vs. Order Amount", fontsize=14, fontweight="bold")
plt.xlabel("Unit Price ($)")
plt.ylabel("Order Amount ($)")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))
sns.boxplot(
    data=df,
    x="Warehouse_Region",
    y="Order_Amount",
    hue="Returned",
    palette="PRGn",  # Using a diverging palette
    linewidth=1.5,
    fliersize=4,      # Slightly smaller outliers
)
plt.title(
    "Order Amount by Warehouse Region and Return Status",
    fontsize=16,
    fontweight="bold",
)
plt.xlabel("Warehouse Region", fontsize=13)
plt.ylabel("Order Amount ($)", fontsize=13)
plt.legend(title="Returned?", title_fontsize=12)
plt.tight_layout()
plt.show()
```
