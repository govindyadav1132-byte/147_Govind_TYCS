```python
data ={
    "city":["Mumbai","Delhi","Bangalore","Chennai","Kolkata"],
    "temperature_c": [30.5,28.2,32.1,27.8,29.9],
    "humidity_pct": [78,45,65,80,82],
    "is_coastal":[True,True,False,False,True]
    }

df= pd.DataFrame(data)
df
```

```python
print("Shape (rows , cols):", df.shape)
print("\nColumn names :",list(df.columns))
print(df.dtypes)
print("\nQuick Summary stats :")
df.describe()
```

```python
print("Row with label 2 (via .loc):")
print(df.loc[2])
print("\n")
print("\n")

print("n First 2 rows, first 2 col;umns (via .iloc) :")
print(df.iloc[0:2 , 0:2])
print("\n")
print("\n")

print("\n Just the city column(return a Series) :")
print(df["city"])

print("\nSpecific cell: city of rows 3 (via .loc) :")
print(df.loc[3, 'city'])
```

```python
hot= df[df["temperature_c"] > 33 ]
print(" Cities hotter than 33 C")

print(hot)

coastal_and_humid = df[(df["is_coastal"] == True) & (df["humidity_pct"] > 75)]
print("\n Coastal cities with hmidity > 75% :")
print(coastal_and_humid)
```
